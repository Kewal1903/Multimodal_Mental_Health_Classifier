### **Experiment 1**: Vacuum Cleaner Agent<br>
---
A vacuum cleaner agent operates in a simple environment consisting of two rooms, labeled A and B, arranged linearly such that room A is to the left of room B. Each room can be in one of two states: Clean or Dirty.

The agent is located in exactly one room at any time and is equipped with sensors that allow it to
perceive:
- its current location (A or B), and
- the cleanliness status (Clean or Dirty) of the current room only.

The agent can perform the following actions:
- Suck: cleans the current room if it is dirty,
- Left: moves the agent from room B to room A,
- Right: moves the agent from room A to room B.

The environment is assumed to be fully observable, deterministic, static, discrete, and single-agent.

The goal of the agent is to clean all rooms while minimizing unnecessary movements.

PEAS Description:
- Performance: Cleanliness, minimal movement
- Environment: Two rooms
- Actuators: Left, Right, Suck
- Sensors: Location, Dirt

```python
class VacuumAgent:
  def __init__(self):
    self.location = 'A' # defining a default location

  def act(self, percept):
    '''
    - percept is a tuple containing location and status like ('A', 'Dirty')
    - the agent sucks if there is dirt at the landing location
    - the agent moves left/right (whichever feasible) if the landing loc is clean
    '''
    location, status = percept
    if location == 'A':
      if status == 'Dirty':
        print(f"Sucking dirt at A")
      else:
        print(f"A is clean, moving right to B")
    elif location  == 'B':
      if status == 'Dirty':
        print(f"Sucking dirt at B")
      else:
        print(f"B is clean, moving left to A")
```

```python
v = VacuumAgent()
v.act(('A', 'Dirty'))
v.act(('A', 'Clean'))
v.act(('B', 'Dirty'))
v.act(('B', 'Clean'))
```

### Experiment 1: Breadth-First Search (BFS) for Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent is placed at a start node and must reach a goal node by traversing the maze.

**Task**:
1. Formulate the maze navigation problem as a search problem by clearly defining:
    - State space
    - Initial state
    - Actions
    - Transition model
    - Goal test
    - Path cost
2. Implement the Breadth-First Search (BFS) algorithm to find the shortest path from the start node to the goal node.
3. Print the path returned by BFS.
4. Count and display the number of nodes expanded during the search.

**Constraints**:
- The maze is unweighted.
- The environment is fully observable, deterministic, static, discrete, and single-agent.
- BFS must be implemented without using any built-in graph search libraries.

```python
from collections import deque

maze1 = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': []
}

start_node = 'A'
goal_node ='G'

def bfs_maze(start, goal, maze):
  # Create a deque.
  # The deque will store full paths, not just nodes
  dq = deque()
  dq.append([start]) # Push the current node to the deque

  # Define a set to store visited nodes to avoid revisiting them
  visited = set()

  # Define a counter to keep track of no. of nodes expanded
  nodes_expanded = 0

  # Iteratively, pick items (lists) from the left of the deque and process them
  # Popping from left is same as the typical dequeue in case of a typical queue
  # Iterations need to happen till deque is empty
  while dq:
    # Extract the first-most (left-most) path
    path = dq.popleft()
    # Get the node at which the agent is currently at - last node of path
    current_node = path[-1]

    # Check if the current position of agent is the goal, if yes return the path
    if current_node == goal:
      return path, nodes_expanded

    # Expand the node only if it has not already been visited
    if current_node not in visited:
      visited.add(current_node)
      nodes_expanded += 1

      # Fetch all the neighbouring nodes of the current node from the maze
      # Iterative over all the neighbours and attach them to the existing path
      # Create new path for each append and push the paths to the deque
      for neighbour in maze[current_node]:
        # path is already a list, this is a sanity step
        new_path = list(path)
        # append the neighbour to the path that we already have
        new_path.append(neighbour)
        # push the new path to the deque
        dq.append(new_path)

  # The code should not reach this point
  return None, nodes_expanded

path, expanded = bfs_maze(start_node, goal_node, maze1)

print(f"Path: {path}")
print(f"Nodes Expanded: {expanded}")
```

### Experiment 2: Depth-First Search (DFS) for Graph Traversal
**Problem Statement**:<br>
Given a connected, finite graph represented using an adjacency list, design and implement the Depth-First Search (DFS) algorithm to find a path from a specified start node to a goal node.

**Task Requirements**:<br>
1. Formulate the graph traversal problem by identifying:
    - State representation
    - Initial state
    - Goal state
    - Actions
2. Implement a recursive DFS algorithm that:
    - Explores one branch of the search tree as deeply as possible before backtracking
    - Avoids revisiting nodes already present in the current path

3. Return and display:
    - A valid path from the start node to the goal node (if it exists)

**Constraints**<br>
- The graph is unweighted.
- DFS must be implemented without using any built-in graph traversal libraries.
- The algorithm is not required to return the shortest path.

**Expected Outcome**:<br>
The algorithm should successfully find a path from the start node to the goal node by exploring depth-
wise, though the path may not be optimal.

```python
# The question demands a recursive approach
# The fxn should take in the current node, and the path and the goal node
# At each step, we call the fxn recursively for the nighbours of the curr node

from collections import deque

maze2 = {
    'A': ['B', 'C'],
    'B': ['A', 'D', 'E'],
    'C': ['A', 'F'],
    'D': ['B'],
    'E': ['B', 'F'],
    'F': ['C', 'E', 'G'],
    'G': []
}

start_node = 'A'
goal_node = 'G'
visited = set()
path_ = []

def dfs_maze(maze, start, goal, path, visited):
  # Attach the cuurent start node to the path (list) and visited (set)
  path.append(start)
  visited.add(start)

  # Check if the current node is the goal
  if start == goal:
    return path

  # If the curr node is not goal, iteratively call the fxn on all its neighbours
  for neighbour in maze[start]:
    # Process the neighbour only if hasn't already been visited
    if neighbour not in visited:
      # Store the result as it is needs to travel back during stack unwinding
      # The result is a path (list)
      result = dfs_maze(maze, neighbour, goal, path, visited)

      # Returning the result to the parent call to make it travel back
      if result is not None:
        return result

  # The leaf (end-most) nodes that arent the goal node will reach here
  # Since its a leaf node, we can't travel to any other node from it
  # And it isn't the goal as well, meaning it should not be a part of the path
  # Hence we pop it from the `path` list
  # Other (non-leaf) nodes will never get to this point
  # Those nodes will trigger one of the above `return` statements and unwind
  path.pop()

  # Since leaf nodes don't have anything meaningful to propagate, return None
  return None

soln = dfs_maze(maze2, start_node, goal_node, path_, visited)

if soln:
  print(f"Path discovered: {soln}")
else:
  print(f"No path discovered")
```

### Experiment 3: Solving the 8-Puzzle Problem Using BFS
**Problem Statement**:<br>
The 8-puzzle problem consists of a 3×3 grid containing eight numbered tiles and one blank space. A tile adjacent to the blank space can be moved into the blank position. Given an initial configuration and a goal configuration, implement Breadth-First Search (BFS) to find a sequence of moves that transforms the initial state into the goal state.

**Task Requirements**:
1. Represent the puzzle state using a suitable data structure.
2. Define:
    - Initial state
    - Goal state
    - Valid actions (up, down, left, right)
    - Transition model
3. Implement BFS to explore the state space.
4. Output:
    - The sequence of states from the initial state to the goal state
    - The total number of states expanded during the search

**Constraints**:
- Use BFS to guarantee the shortest solution.
- Avoid revisiting previously explored states.
- Assume the given initial state is solvable.

**Expected Outcome**:<br>
The algorithm should produce the shortest sequence of moves required to reach the goal configuration.

**Phase 1: The Setup**<br>
- Step 1: <br>
Define States,"Create your initial_state and goal_state as 1D tuples (e.g., (1, 2, 3, 4, 5, 6, 7, 8, 0))."

- Step 2:<br>
Initialize Tracking,"Create an empty visited set to avoid revisiting states, and add your initial_state to it immediately. Set up a `nodes_expanded = 0` counter."

- Step 3:<br>
Initialize Queue,Create a deque() for BFS. Push a list containing just the initial_state into the queue (this is your starting path).
<br>

**Phase 2: The BFS Loop**<br>

- Step 4:<br>
Pop the Path,Start a while queue: loop. Pop the oldest path from the left of the queue.+1

- Step 5:<br>
Extract Current State,Get the current_state by taking the last tuple in the path you just popped (path[-1]).

- Step 6: <br>
Goal Test,"Check if the current_state exactly matches the goal_state. If yes, return the path and nodes_expanded ."

- Step 7:<br>
Expand the Node,"If it is not the goal, increment your nodes_expanded counter by 1. Now, find the index of 0 in the current_state tuple."

- Step 8:<br>
Calculate Valid Moves,"Based on the index of 0, determine which of the 4 moves (Up, Down, Left, Right) are valid using the boundary math we discussed."

- Step 9:<br>
Generate Neighbors,"For every valid move, calculate the new index for 0. Convert the current_state tuple to a list, swap the 0 with the target number, and convert it back into a new_state tuple."

- Step 10: <br>
Filter and Queue,Check if this new_state is already in the visited set. If it is not: <br>
  a) Add it to the visited set. <br>
  b) Make a copy of the current path (list(path)).<br>
  c) Append the new_state to this copied path. <br>
  d) Push the extended path into the queue.

```python
from collections import deque

init_state = (1, 2, 3,
              4, 0, 5,
              6, 7, 8)

goal_state = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

def get_valid_moves(idx):
  moveset = []
  # UP move can be performed as long as agent is not in the top row
  # UP movement is done by current index - 3
  if idx > 2:
    moveset.append(-3) # appending the value to add for new pos

  # DOWN move can be performed as long as the agent is not in the btm row
  # DOWN movement is done by current index + 3
  if idx < 6:
    moveset.append(3)

  # LEFT is possible as long as the agent is not on one of the left-most indices
  # LEFT movement is done by index - 1
  if idx not in [0, 3, 6]:
    moveset.append(-1)

  # RIGHT is possible as long as the agent is not on one of the rt-most indices
  # RIGHT movement is done by index + 1
  if idx not in [2, 5, 8]:
    moveset.append(1)

  return moveset

def bfs_8_puzzle(start, goal):
  # Instantiate a set
  visited = set()
  # Instantiate a deque
  dq = deque()
  # Append the start state to the deque - pass it encapsulated in a list
  # We will store the entire history of states in a list in the deque,
  dq.append([start])
  # Keep a track of nodes expanded
  nodes_expanded = 0

  # Iteratively handle the nodes and arrangements inside a loop
  while dq:
    # Fetch the left-most arrangement list
    state_list = dq.popleft()
    # Fetch the current arrangement from the current state list
    curr_state = state_list[-1]

    # Check if the current state is the goal state
    if curr_state == goal:
      return state_list, nodes_expanded

    # Check if the state has already been visited - if not, process it
    if curr_state not in visited:
      # Add the state to visited
      visited.add(curr_state)
      nodes_expanded += 1

      # Fetch the index of 0 in the curr_state tuple
      zero_idx = curr_state.index(0)
      # Get the list of valid moves that can be performed
      moves = get_valid_moves(zero_idx)

      # Iteratively perform all moves
      # To perform a move, swap the element at the new index with 0
      # In order to perform the swap, the tuple needs to be converted to a list
      # Then swaps are performed on the list
      # And then list is converted back to a tuple
      # This new tuple is then appended to the state_list i.e. history of moves
      for move in moves:
        # Convert tuple to list
        new_state = list(curr_state)
        # Calc the index with which 0 will be swapped
        new_idx = zero_idx + int(move)
        # Perform swap
        new_state[new_idx], new_state[zero_idx] = new_state[zero_idx], new_state[new_idx]
        # Convert the list back to a tuple
        new_state = tuple(new_state)
        # Make a new history list - a copy of the current one
        new_state_list = list(state_list)
        # Append the new state tuple to the new history list
        new_state_list.append(new_state)
        # Push this new/edited state_list into the deque -
        # DO NOT FORGET - WITHOUT THIS LOOP WILL RUN ONLY ONCE
        dq.append(new_state_list)

  return state_list, nodes_expanded

history, nodes_expd = bfs_8_puzzle(init_state, goal_state)

if history:
  print(f"Path:")
  for i in history:
    print(f"{i}")
  print(f"Nodes expanded: {nodes_expd}")
else:
  print(f"No possible path")
```

### Experiment 1: Greedy Best-First Search for Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent is placed at a start node and must reach a goal node by traversing the maze. **Constraints**:<br>
- The maze is unweighted.

#### Greedy Best-First Search (GBFS) Implementation Idea

**Core Concept:** GBFS is an "informed" search algorithm that chooses the next node based purely on a heuristic estimate, $h(n)$, which predicts how close it is to the goal. It is "greedy" because it ignores the cost of the path taken so far and always chooses the node that *looks* closest to the finish line.

**Key Components:**
* **Heuristic ($h(n)$):** In grid or maze problems, this is typically the Manhattan distance: $|x_1 - x_2| + |y_1 - y_2|$.
* **Data Structure:** A Priority Queue (implemented via Python's `heapq`).

**Implementation Workflow:**
1.  **Initialize:** Push the starting node into the priority queue as a tuple: `(heuristic_value, [start_node_path])`.
2.  **Loop:** While the queue is not empty, pop the path with the lowest heuristic value.
3.  **Goal Test:** If the last node in the extracted path is the goal, return the path.
4.  **Expand:** For each unvisited neighbor:
    * Calculate its $h(n)$ (Manhattan distance to the goal).
    * Append the neighbor to a copy of the current path.
    * Push the new tuple `(neighbor_h, new_path)` into the priority queue.

| Feature | GBFS Characteristic |
| :--- | :--- |
| **Sorting Metric** | $h(n)$ only. |
| **Optimality** | Not guaranteed to find the shortest path. |
| **Efficiency** | Expands fewer nodes than BFS but can get trapped in dead ends. |

```python
import heapq

# 1. The Adjacency List (Topology)
maze3 = {
    'S': ['A', 'B'],
    'A': ['S', 'C', 'D'],
    'B': ['S', 'D', 'E'],
    'C': ['A'],
    'D': ['A', 'B', 'G'],
    'E': ['B'],
    'G': ['D']
}

# 2. The Coordinate Dictionary (Geometry - (x, y) coordinates)
maze3_coords = {
    'S': (0, 0),
    'A': (1, 0),
    'B': (0, 1),
    'C': (2, 0),
    'D': (1, 1),
    'E': (0, 2),
    'G': (2, 2)
}

start_node = 'S'
goal_node = 'G'
visited = set()

def manh_dist(curr, goal):
  xs, ys = maze3_coords[curr]
  xg, yg = maze3_coords[goal]

  return abs(xg - xs) + abs(yg - ys)

def gbfs_maze(maze, coords, start, goal):
  # Get the manh dist for the start node to push it into the heap
  start_manh = manh_dist(start, goal)
  # Create an empty list to serve as the container in the priority queue
  pq = []
  # Instantiate the priority queue - make sure to use the corrrect tuple order
  # Priority queue sorts items based on the first item of the tuples inserted
  heapq.heappush(pq, (start_manh, [start]))

  # Keep a counter for nodes expanded
  nodes_expanded = 0

  # Iteratively handle the items inside a loop
  while pq:
    # Get the best node i.e. node with least manh dist
    best_md, best_path = heapq.heappop(pq)
    # Get the node the agent is currently at
    curr_node =  best_path[-1]

    # Return path if curr node is goal
    if curr_node == goal:
      return nodes_expanded, best_path

    # Check if visted, update otherwise
    if curr_node not in visited:
      visited.add(curr_node)
      nodes_expanded += 1

      # Iteratively visit all neighbours of curr node
      # Calculate the manh dist for all of them
      # Append the neighbour node to the path list
      # Push the tuple (manh dist, new_path_list) to the pq
      for neighbour in maze[curr_node]:
        new_path = list(best_path)
        # Append the neighour to the new path list, NOT TO THE ORIGINAL PATH
        new_path.append(neighbour)
        # Calculate the heuristic for the neighbour
        neighbour_manh = manh_dist(neighbour, goal)
        # Push the tuple (manh_dist, new_path_list) to the priority queue
        heapq.heappush(pq, (neighbour_manh, new_path))

  # The code should not reach this part
  return nodes_expanded, best_path


nodes_expd, soln_path = gbfs_maze(maze3, maze3_coords, start_node, goal_node)

if soln_path:
  print(f"Path: {soln_path}, Nodes expanded: {nodes_expd}")
else:
  print(f"No Path found")
```

### Experiment 2: A* Search for Optimal Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent is placed at a start node and must reach a goal node. To ensure optimal navigation, the agent will use the A* search algorithm, which combines:<br>
    Actual cost from start 𝑔(𝑛) and Heuristic estimate to goal ℎ(𝑛)

        $h(n)$ = Manhattan distance from node n to goal
4. Print the path returned by A* Search.

**Constraints**:<br>

#### A* Search Implementation Idea

**Core Concept:**
$A^*$ is an optimal informed search algorithm. It improves upon GBFS by tracking both the actual cost incurred so far and the estimated cost to the goal. It evaluates nodes by combining these two values: $f(n) = g(n) + h(n)$

**Key Components:**

| Component | Definition | Implementation |
| :--- | :--- | :--- |
| **$g(n)$** | Actual path cost from the start to node $n$. | `len(current_path) - 1` (assuming a uniform step cost of $1$). |
| **$h(n)$** | Heuristic estimate from node $n$ to the goal. | Manhattan distance. The heuristic must be admissible, meaning it never overestimates the true cost. |
| **$f(n)$** | Total estimated cost of the path through node $n$. | The final value used to sort items in the priority queue. |

**Implementation Workflow:**
1.  **Initialize:** Push the starting node into the priority queue as a tuple: `(f_value, [start_node_path])`. For the start node, $g(n) = 0$, so $f(n) = h(n)$.
2.  **Loop:** While the queue is not empty, pop the path with the lowest $f(n)$ value.
3.  **Goal Test:** If the last node in the extracted path is the goal, return the path.
4.  **Expand:** For each unvisited neighbor:
    * Create a `new_path` by appending the neighbor.
    * Calculate $g(n) = \text{len}(\text{new\_path}) - 1$.
    * Calculate $h(n) = \text{Manhattan distance to goal}$.
    * Calculate $f(n) = g(n) + h(n)$.
    * Push the new tuple `(f_n, new_path)` into the priority queue.

```python
import heapq

# 1. The Adjacency List (Topology)
maze4 = {
    'S': ['A', 'B'],
    'A': ['S', 'C', 'D'],
    'B': ['S', 'D', 'E'],
    'C': ['A'],
    'D': ['A', 'B', 'G'],
    'E': ['B'],
    'G': ['D']
}

# 2. The Coordinate Dictionary (Geometry - (x, y) coordinates)
maze4_coords = {
    'S': (0, 0),
    'A': (1, 0),
    'B': (0, 1),
    'C': (2, 0),
    'D': (1, 1),
    'E': (0, 2),
    'G': (2, 2)
}

start_node = 'S'
goal_node = 'G'
visited = set()

def manh_dist(curr, goal):
  xs, ys = maze4_coords[curr]
  xg, yg = maze4_coords[goal]

  return abs(xg - xs) + abs(yg - ys)

def a_star_maze(start, goal, maze, coords):
  # Get the manh dist for the start node
  start_manh = manh_dist(start, goal)
  # Add the start node to a list
  start_path = [start]
  # Make a list to be used as pq
  pq = []
  # Instantiate a priority queue and push the first record into it
  # h(n) = manhattan distance = start_manh
  # g(n) = length of curr path list - 1 = len(start_path) - 1
  # f(n) = g(n) + h(n)
  heapq.heappush(pq, (start_manh + len(start_path) - 1, start_path))
  # Make a nodes expanded variable
  nodes_expanded = 0

  # Iteratively handle the nodes
  while pq:
    # Get the top most element - element with lowest f(n)
    f_n, curr_path = heapq.heappop(pq)
    # Get the curret node from the curr path - last node of the path
    curr_node = curr_path[-1]

    # Check if the current node is the goal
    if curr_node == goal:
      return nodes_expanded, curr_path

    if curr_node not in visited:
      visited.add(curr_node)
      nodes_expanded += 1

    # Iteratively process all neighbours
    # Calc the f(n) value for all of them
    # Attach the neigbours to the current path
    # Make tuples - (f(n), new_path) and push them to pq

    for neighbour in maze[curr_node]:
      if neighbour not in visited:
        new_path = list(curr_path)
        # Append the neighbour to the path
        new_path.append(neighbour)
        # Calculate the f(n) value for the neighbour
        f_n_neigh = manh_dist(neighbour, goal) + len(new_path) - 1
        # Make and push the tuple to pq
        heapq.heappush(pq, (f_n_neigh, new_path))

  return nodes_expanded, None

nodes_expd, soln = a_star_maze(start_node, goal_node, maze4, maze4_coords)

if soln:
  print(f"Path: {soln}, Nodes expanded: {nodes_expd}")
else:
  print(f"No Path found")
```

### Experiment 1: Recursive Best-First Search (RBFS) for Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent is placed at a start node and must reach a goal node by traversing the maze. Traditional A* requires storing all explored nodes in memory, which may become infeasible for large mazes. Recursive Best-First Search (RBFS) is a memory-bounded heuristic search that uses recursion and limited memory while still aiming for optimal solutions.

**Task**:<br>
1. Formulate the maze navigation problem as a search problem by clearly defining: State space, Initial state, Actions, Transition model, Goal test, Path cost, Heuristic function ℎ(𝑛)
2. Implement the Recursive Best-First Search (RBFS) algorithm to find an optimal path from the start node to the goal node.
3. Use the evaluation function: $𝑓(𝑛)=𝑔(𝑛)+ℎ(𝑛)$ where:<br>
    $g(𝑛)$ = path cost from starts to node 𝑛<br>
    $ℎ(𝑛)$ = Manhattan distance from node 𝑛 to the goal
4. Print the path returned by RBFS.
5. Count and display the number of nodes expanded during the search.

**Constraints**:<br>
- The maze is unweighted (each step cost = 1).
- The environment is fully observable, deterministic, static, discrete, and single agent.
- RBFS must be implemented without using built-in graph search libraries.
- The algorithm should use only linear memory through recursion.

```python
# ==========================================
# EXPERIMENT 1: RBFS MAZE NAVIGATION
# ==========================================

# 1. The Adjacency List (Topology)
maze = {
    'S': ['A', 'B'],
    'A': ['S', 'C', 'D'],
    'B': ['S', 'D', 'E'],
    'C': ['A'],
    'D': ['A', 'B', 'G'],
    'E': ['B'],
    'G': ['D']
}

# 2. The Coordinate Dictionary (Geometry)
maze_coords = {
    'S': (0, 0),
    'A': (1, 0),
    'B': (0, 1),
    'C': (2, 0),
    'D': (1, 1),
    'E': (0, 2),
    'G': (2, 2)
}

# Helper function: Manhattan Distance
def manh_dist(curr, goal, coords):
    xs, ys = coords[curr]
    xg, yg = coords[goal]
    return abs(xg - xs) + abs(yg - ys)

def rbfs_maze(maze, coords, start, goal):
    # We use a list for nodes_expanded so it can be mutated inside the recursive function
    # without running into Python "global" variable scope issues.
    nodes_expanded = [0]

    # ==========================================
    # THE RECURSIVE HELPER FUNCTION
    # ==========================================
    def rbfs(path, current_f, f_limit):
        current_node = path[-1]

        # STEP 1: Goal Test
        if current_node == goal:
            return path, current_f

        # STEP 2: Get Neighbors
        # We filter out nodes already in our CURRENT path to prevent inf loops,
        # but we don't use a global 'visited' set because RBFS needs to be able
        # to forget and re-explore paths to save memory).
        neighbors = [n for n in maze[current_node] if n not in path]

        # STEP 3: Dead End Test
        if not neighbors:
            # Return None for path, and Inf for cost so we never come here again
            return None, float('inf')

        # We are officially expanding this node!
        nodes_expanded[0] += 1

        # STEP 4 & 5: Generate Successors and Calculate F-Values
        successors = []
        for neighbor in neighbors:
            new_path = list(path)
            new_path.append(neighbor)

            g_n = len(new_path) - 1 # Step cost is 1
            h_n = manh_dist(neighbor, goal, coords)
            raw_f = g_n + h_n

            # THE RBFS RULE: A child's f-value cannot be lower than its parent's!
            child_f = max(raw_f, current_f)

            # STEP 6: Store Successors
            # We store it as a list [f_value, path] so we can update the f_value later
            successors.append([child_f, new_path])

        # ==========================================
        # STEP 7-12: THE EVALUATION LOOP
        # ==========================================
        while True:
            # STEP 7: Sort successors by their f-value (index 0)
            successors.sort(key=lambda x: x[0])

            best_f = successors[0][0]
            best_path = successors[0][1]

            # STEP 8: Check Limit (Backtrack if current path is too expensive!) [cite: 353]
            if best_f > f_limit:
                return None, best_f

            # STEP 9: Find Alternative (The second best option) [cite: 352]
            if len(successors) > 1:
                alt_f = successors[1][0]
            else:
                alt_f = float('inf') # If no alternative, the alternative cost is infinite

            # STEP 10: Set New Limit
            # The new budget is the strict minimum of our current limit and the alternative
            new_limit = min(f_limit, alt_f)

            # STEP 11: Recurse! Dive into the best path
            result, updated_f = rbfs(best_path, best_f, new_limit)

            # STEP 12: Handle the Result
            if result is not None:
                # We found the goal! Pass the successful path all the way back up.
                return result, updated_f
            else:
                # We hit a dead end or exceeded the limit.
                # Update this node's f-value with the backed-up cost.
                # The while loop will restart, re-sort the list, and try the NEW best option!
                successors[0][0] = updated_f

    # ==========================================
    # INITIAL FUNCTION EXECUTION
    # ==========================================
    initial_h = manh_dist(start, goal, coords)
    # The first call has an f_limit of Infinity, meaning it is free to explore anywhere
    final_path, final_f = rbfs([start], initial_h, float('inf'))

    return final_path, nodes_expanded[0]

# Run the algorithm
start_node = 'S'
goal_node = 'G'
path, expanded = rbfs_maze(maze, maze_coords, start_node, goal_node)

if path:
    print(f"Path: {path}")
    print(f"Nodes expanded: {expanded}")
else:
    print("No Path found")
```

### Experiment 2: Recursive Best-First Search (RBFS) for 8-Puzzle problem
**Problem Statement**:<br>
Consider the 8-Puzzle problem, where tiles are arranged in a 3 × 3 grid containing eight numbered tiles and one blank space. A tile adjacent to the blank space can be moved into the blank position. An intelligent agent must transform the initial puzzle configuration into the goal configuration using an optimal strategy.
Traditional A* search may run out of memory when the search space becomes large. Recursive Best-First Search (RBFS) is a memory-efficient variant of A* that works by:
- Expanding the best node first
- Using recursion instead of storing all generated nodes
- Backtracking when alternative paths have better evaluation cost
- Maintaining optimality when an admissible heuristic is used

**Task**:<br>
1. Formulate the maze navigation problem as a search problem by clearly defining: State space, Initial state, Actions, Transition model, Goal test, Path cost, Heuristic function $h(n)$
2. Implement the Recursive Best-First Search (RBFS) algorithm to find the optimal sequence of moves from the start state to the goal state.
3. Use the evaluation function:
        $f(n) = g(n) + h(n)$
4. During node expansion:
    - Maintain the best alternative f-value
    - Backtrack when current node exceeds alternative path cost
5. Print the final sequence of puzzle states returned by RBFS.
6. Count and display the number of nodes expanded during the search.

**Constraints**:<br>
- Uniform step cost = 1
- Heuristic must be admissible (Use Manhattan Distance)
- Environment is fully observable, deterministic, static, discrete, and single-agent
- RBFS must be implemented without built-in search libraries
- The algorithm must minimize memory usage

```python
# ==========================================
# EXPERIMENT 2: RBFS 8-PUZZLE PROBLEM
# ==========================================

init_state = (1, 2, 3,
              4, 0, 5,
              6, 7, 8)

goal_state = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

# Helper 1: Calculate Valid Moves for the Blank Tile (0)
def get_valid_moves(idx):
    moveset = []
    if idx > 2: moveset.append(-3)              # UP
    if idx < 6: moveset.append(3)               # DOWN
    if idx not in [0, 3, 6]: moveset.append(-1) # LEFT
    if idx not in [2, 5, 8]: moveset.append(1)  # RIGHT
    return moveset

# Helper 2: Manhattan Distance for 8-Puzzle (Admissible Heuristic)
def manh_dist_8_puzzle(state, goal):
    dist = 0
    # Calculate distance for tiles 1 through 8 (ignore the 0)
    for tile in range(1, 9):
        curr_idx = state.index(tile)
        goal_idx = goal.index(tile)

        # Convert 1D index to 2D (row, col) coordinates
        curr_row, curr_col = curr_idx // 3, curr_idx % 3
        goal_row, goal_col = goal_idx // 3, goal_idx % 3

        dist += abs(curr_row - goal_row) + abs(curr_col - goal_col)
    return dist

# Main Algorithm: RBFS for 8-Puzzle
def rbfs_8_puzzle(start, goal):
    nodes_expanded = [0]

    # ==========================================
    # THE RECURSIVE HELPER FUNCTION
    # ==========================================
    def rbfs(path, current_f, f_limit):
        curr_state = path[-1]

        # STEP 1: Goal Test
        if curr_state == goal:
            return path, current_f

        # STEP 2: Generate Successors (Valid Neighbors)
        zero_idx = curr_state.index(0)
        moves = get_valid_moves(zero_idx)

        neighbors = []
        for move in moves:
            new_idx = zero_idx + move
            new_state_list = list(curr_state)
            # Swap the 0 with the target tile
            new_state_list[zero_idx], new_state_list[new_idx] = new_state_list[new_idx], new_state_list[zero_idx]
            new_state = tuple(new_state_list)

            # Cycle Check: Only add if not already in our current path
            if new_state not in path:
                neighbors.append(new_state)

        # STEP 3: Dead End Test
        if not neighbors:
            return None, float('inf')

        nodes_expanded[0] += 1
        successors = []

        # STEP 4 & 5: Calculate Costs and F-Values
        for neighbor in neighbors:
            new_path = list(path)
            new_path.append(neighbor)

            # Step cost is uniform (= 1)
            g_n = len(new_path) - 1
            h_n = manh_dist_8_puzzle(neighbor, goal)
            raw_f = g_n + h_n

            # Pathmax: A child's f-value cannot be lower than its parent's!
            child_f = max(raw_f, current_f)

            # Store as [f_value, path] so we can mutate the f_value later
            successors.append([child_f, new_path])

        # ==========================================
        # STEP 6: THE EVALUATION LOOP (Backtracking)
        # ==========================================
        while True:
            # Sort successors by their f-value (index 0) to get the best node first
            successors.sort(key=lambda x: x[0])

            best_f = successors[0][0]
            best_path = successors[0][1]

            # Check Limit: If the best path exceeds our alternative limit, backtrack!
            if best_f > f_limit:
                return None, best_f

            # Find Alternative: The f-value of the second-best option
            if len(successors) > 1:
                alt_f = successors[1][0]
            else:
                alt_f = float('inf')

            # Set New Limit: The minimum of our current limit and the alternative
            new_limit = min(f_limit, alt_f)

            # Recurse: Dive into the best path!
            result, updated_f = rbfs(best_path, best_f, new_limit)

            # Handle the Result
            if result is not None:
                # Goal found! Bubble the path back up.
                return result, updated_f
            else:
                # We hit a dead end or exceeded the f_limit.
                # Update this node's f-value and let the while loop re-sort!
                successors[0][0] = updated_f

    # ==========================================
    # INITIAL FUNCTION EXECUTION
    # ==========================================
    initial_h = manh_dist_8_puzzle(start, goal)
    # The starting f_limit is Infinity
    final_path, final_f = rbfs([start], initial_h, float('inf'))

    return final_path, nodes_expanded[0]

# Run the algorithm
path, expanded = rbfs_8_puzzle(init_state, goal_state)

if path:
    print(f"Path length: {len(path) - 1} steps")
    print("Path sequence:")
    for step, state in enumerate(path):
        print(f"Step {step}:\n{state[0:3]}\n{state[3:6]}\n{state[6:9]}\n")
    print(f"Total Nodes dynamically expanded: {expanded}")
else:
    print("No Path found.")
```

### Experiment 3: Iterative Deepening A* (IDA*) for Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent must reach the goal node efficiently. While A* guarantees optimality, it requires large memory. #### Iterative Deepening $A^*$ ($IDA^*$) Implementation Idea

**Core Concept:** $IDA^*$ is a memory-bounded heuristic search that combines the memory efficiency of Depth-First Search (DFS) with the optimal pathfinding of $A^*$. Instead of limiting how *deep* a search can go, it limits how *expensive* the search can get by using a strict cost threshold. It performs repeated depth-first searches with increasing cost thresholds based on the $A^*$ evaluation function.

**Key Components:**

| Component | Definition | Role in $IDA^*$ |
| :--- | :--- | :--- |
| **$f(n)$** | Evaluation function: $f(n) = g(n) + h(n)$ | Calculates the total estimated cost of a path. |
| **Threshold** | The current $f$-cost limit. | Acts as the strict "budget" for the current DFS iteration. |
| **The Explorer** | A recursive DFS function. | Traverses the graph but immediately stops and backtracks if $f(n) > \text{threshold}$. |
| **The Manager** | An iterative `while` loop. | Launches the Explorer and raises the threshold if the goal isn't found. |

**Implementation Workflow:**
1.  **Initialize:** Set the starting threshold to the heuristic value of the start node: $h(\text{start})$.
2.  **DFS Dive:** Send the recursive DFS to explore. For every node, calculate $f(n) = g(n) + h(n)$.
3.  **Cut-Off Test:** If a node's $f(n)$ exceeds the threshold, immediately return `None` (for the path) and its $f(n)$ value.
4.  **Track Minimum Exceeded Cost:** As the DFS unwinds from cut-offs, it tracks the *absolute lowest* $f$-cost that exceeded the current budget.
5.  **Iterate and Deepen:** If the DFS finishes without finding the goal, the Manager updates the threshold to that lowest exceeded cost and restarts the DFS from the beginning.

**Characteristics:**
* **Memory:** Extremely low (linear memory), as it only remembers the current path being explored.
* **Optimality:** Guaranteed to find the optimal path if the heuristic is admissible.
* **Trade-off:** Requires less memory but may increase computation time because it re-expands the shallow nodes multiple times during each iteration.

```python
# ==========================================
# EXPERIMENT 3: IDA* MAZE NAVIGATION
# ==========================================

# 1. The Adjacency List (Topology)
maze = {
    'S': ['A', 'B'],
    'A': ['S', 'C', 'D'],
    'B': ['S', 'D', 'E'],
    'C': ['A'],
    'D': ['A', 'B', 'G'],
    'E': ['B'],
    'G': ['D']
}

# 2. The Coordinate Dictionary (Geometry)
maze_coords = {
    'S': (0, 0),
    'A': (1, 0),
    'B': (0, 1),
    'C': (2, 0),
    'D': (1, 1),
    'E': (0, 2),
    'G': (2, 2)
}

# Helper function: Manhattan Distance
def manh_dist(curr, goal, coords):
    xs, ys = coords[curr]
    xg, yg = coords[goal]
    return abs(xg - xs) + abs(yg - ys)

def ida_star_maze(maze, coords, start, goal):
    # Use a list for the counter so the recursive function can modify it
    nodes_expanded = [0]

    # ==========================================
    # PHASE 2: THE EXPLORER (Recursive DFS) [cite: 379]
    # ==========================================
    def dfs_contour(path, g_cost, threshold):
        current_node = path[-1]

        # Step 1: Calculate Current Cost [cite: 370]
        h_cost = manh_dist(current_node, goal, coords)
        f_cost = g_cost + h_cost

        # Step 2: The Cut-off Test
        # If this node exceeds our current budget, STOP and return the cost
        if f_cost > threshold:
            return None, f_cost

        # Step 3: The Goal Test
        if current_node == goal:
            return path, f_cost

        # Step 4: Prepare to Expand
        nodes_expanded[0] += 1
        min_cost = float('inf') # Track the lowest cost of dead-end branches

        # Step 5: Loop Neighbors
        for neighbor in maze[current_node]:
            # Step 6: Cycle Check (Don't go backwards!)
            if neighbor not in path:
                new_path = list(path)
                new_path.append(neighbor)

                # Step 7: Recurse! (Step cost is 1, so g_cost + 1) [cite: 375]
                res_path, returned_cost = dfs_contour(new_path, g_cost + 1, threshold)

                # Step 8: Handle Return Values
                if res_path is not None:
                    # Goal found deeper down! Pass it up.
                    return res_path, returned_cost

                # If it was a dead end or exceeded budget, track the minimum exceeded cost
                min_cost = min(min_cost, returned_cost)

        # Step 9: End of the Line
        # If all neighbors failed, return the lowest exceeded cost we found
        return None, min_cost

    # ==========================================
    # PHASE 1: THE MANAGER (Iterative Deepening Loop)
    # ==========================================
    # Step 2: Set Initial Budget
    initial_threshold = manh_dist(start, goal, coords)

    # Step 3: The Deepening Loop [cite: 371]
    while True:
        # Step 4: Send the Explorer
        result_path, min_exceeded_cost = dfs_contour([start], 0, initial_threshold)

        # Step 5: Evaluate Success
        if result_path is not None:
            return result_path, nodes_expanded[0]

        # Step 6: Evaluate Failure (Graph exhausted)
        if min_exceeded_cost == float('inf'):
            return None, nodes_expanded[0]

        # Step 7: Raise the Budget
        # Set the budget to the exact lowest cost that was cut off in the last search
        initial_threshold = min_exceeded_cost


# Run the algorithm
start_node = 'S'
goal_node = 'G'
path, expanded = ida_star_maze(maze, maze_coords, start_node, goal_node)

if path:
    print(f"Path: {path}")
    print(f"Nodes expanded: {expanded}")
else:
    print("No Path found")
```

```python
# ==========================================
# EXPERIMENT 4: IDA* 8-PUZZLE PROBLEM
# ==========================================

init_state = (1, 2, 3,
              4, 0, 5,
              6, 7, 8)

goal_state = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

# Helper 1: Calculate Valid Moves for the Blank Tile (0)
def get_valid_moves(idx):
    moveset = []
    # UP: current index - 3
    if idx > 2:
        moveset.append(-3)
    # DOWN: current index + 3
    if idx < 6:
        moveset.append(3)
    # LEFT: index - 1
    if idx not in [0, 3, 6]:
        moveset.append(-1)
    # RIGHT: index + 1
    if idx not in [2, 5, 8]:
        moveset.append(1)
    return moveset

# Helper 2: Manhattan Distance for 8-Puzzle
def manh_dist_8_puzzle(state, goal):
    dist = 0
    # Calculate distance for tiles 1 through 8 (ignore the 0)
    for tile in range(1, 9):
        curr_idx = state.index(tile)
        goal_idx = goal.index(tile)

        # Convert 1D index to 2D (row, col) coordinates
        curr_row, curr_col = curr_idx // 3, curr_idx % 3
        goal_row, goal_col = goal_idx // 3, goal_idx % 3

        dist += abs(curr_row - goal_row) + abs(curr_col - goal_col)
    return dist

# Main Algorithm: IDA*
def ida_star_8_puzzle(start, goal):
    nodes_expanded = [0]

    # ==========================================
    # THE EXPLORER: Recursive DFS Contour
    # ==========================================
    def dfs_contour(path, g_cost, threshold):
        curr_state = path[-1]

        # Calculate f(n) = g(n) + h(n)
        h_cost = manh_dist_8_puzzle(curr_state, goal)
        f_cost = g_cost + h_cost

        # Cut-off Test
        if f_cost > threshold:
            return None, f_cost

        # Goal Test
        if curr_state == goal:
            return path, f_cost

        nodes_expanded[0] += 1
        min_cost = float('inf')

        # Generate Successors
        zero_idx = curr_state.index(0)
        moves = get_valid_moves(zero_idx)

        for move in moves:
            # Swap logic to generate new state tuple
            new_idx = zero_idx + move
            new_state_list = list(curr_state)
            new_state_list[new_idx], new_state_list[zero_idx] = new_state_list[zero_idx], new_state_list[new_idx]
            new_state = tuple(new_state_list)

            # Cycle avoidance (don't walk back up the current path)
            if new_state not in path:
                new_path = list(path)
                new_path.append(new_state)

                # Recurse: step cost is strictly 1
                res_path, returned_cost = dfs_contour(new_path, g_cost + 1, threshold)

                if res_path is not None:
                    return res_path, returned_cost

                # Track the lowest cost that exceeded the threshold
                min_cost = min(min_cost, returned_cost)

        return None, min_cost

    # ==========================================
    # THE MANAGER: Iterative Deepening Loop
    # ==========================================
    initial_threshold = manh_dist_8_puzzle(start, goal)

    while True:
        # Launch DFS Explorer
        result_path, min_exceeded_cost = dfs_contour([start], 0, initial_threshold)

        # Success Check
        if result_path is not None:
            return result_path, nodes_expanded[0]

        # Failure Check (Search space exhausted)
        if min_exceeded_cost == float('inf'):
            return None, nodes_expanded[0]

        # Raise Threshold for next iteration
        initial_threshold = min_exceeded_cost

# Run the algorithm
history, expanded = ida_star_8_puzzle(init_state, goal_state)

if history:
    print("Path:")
    for step, state in enumerate(history):
        print(f"Step {step}: {state[0:3]}\n         {state[3:6]}\n         {state[6:9]}\n")
    print(f"Total Nodes Expanded: {expanded}")
else:
    print("No possible path found.")
```

### Experiment 5: Simplified Memory-Bounded A* (SMA*) for Maze Navigation
**Problem Statement**:<br>
Consider a maze represented as a graph, where each node corresponds to a location in the maze and edges represent valid paths between locations. An agent must reach the goal node using an optimal strategy. A* search may run out of memory when the search space becomes large. Simplified Memory-Bounded A* (SMA*) is a variant of A* that uses a fixed memory limit by Expanding best nodes first, Dropping worst leaf nodes when memory is full, retaining optimality when enough memory is available.

5. When memory is full:
    - Remove the node with the highest f(n)f(n)f(n) value
    - Backup its cost to its parent
6. Print the final path returned by SMA*.
7. Count and display the number of nodes expanded during the search.

**Constraints**:<br>

- The heuristic must be admissible.
- The environment is fully observable, deterministic, static, discrete, and single-agent.
- SMA* must be implemented without built-in graph search libraries.
- The algorithm must respect a predefined memory limit.

#### Simplified Memory-Bounded $A^*$ ($SMA^*$) Implementation Idea

**Core Concept:** $SMA^*$ is designed to find optimal solutions while strictly operating under a predefined, fixed memory limit. It behaves exactly like standard $A^*$ until the memory capacity is reached.

**The Pruning and Backup Mechanism:**
When the memory limit is exceeded, $SMA^*$ must make room for new nodes. It achieves this through a specific process:
1.  **Drop the Worst Node:** It identifies the leaf node with the highest evaluation cost, $f(n)$, and deletes it from memory.
2.  **Cost Backup:** To prevent the algorithm from permanently "forgetting" that path, the $f$-value of the deleted node is backed up (assigned) to its parent node.
3.  **Re-evaluation:** If all of a parent's children are deleted, the parent itself becomes a leaf node again, holding the $f$-value of its best forgotten child.

| Feature | $SMA^*$ Characteristic |
| :--- | :--- |
| **Evaluation Function** | $f(n) = g(n) + h(n)$ |
| **Memory usage** | Bounded by a strictly defined maximum capacity. |
| **Optimality** | Optimal *if* the memory limit is large enough to store the shallowest optimal solution path. Otherwise, it will return the best solution it can fit within the memory constraints. |

```python
import heapq

# 1. Our Standard Dictionary Maze
maze = {
    'S': ['A', 'B'],
    'A': ['S', 'C', 'D'],
    'B': ['S', 'D', 'E'],
    'C': ['A'],
    'D': ['A', 'B', 'G'],
    'E': ['B'],
    'G': ['D']
}

maze_coords = {
    'S': (0, 0),
    'A': (1, 0),
    'B': (0, 1),
    'C': (2, 0),
    'D': (1, 1),
    'E': (0, 2),
    'G': (2, 2)
}

start = 'S'
goal = 'G'
MEM_LIMIT = 20
nodes_expanded = 0

# 2. Our standard Manhattan function
def manhattan_maze(curr, goal):
    return abs(maze_coords[curr][0] - maze_coords[goal][0]) + abs(maze_coords[curr][1] - maze_coords[goal][1])

class Node:
    def __init__(self, state, parent, g):
        self.state = state
        self.parent = parent
        self.g = g
        self.f = g + manhattan_maze(state, goal)

    def __lt__(self, other):
        return self.f < other.f

open_list = []
heapq.heappush(open_list, Node(start, None, 0))

while open_list:
    node = heapq.heappop(open_list)
    nodes_expanded += 1

    if node.state == goal:
        path = []
        while node:
            path.append(node.state)
            node = node.parent
        print("Path:", path[::-1])
        print("Nodes Expanded:", nodes_expanded)
        break

    # 3. Use the dictionary to find neighbors
    for n in maze[node.state]:
        # Cycle avoidance (basic)
        if node.parent and n == node.parent.state:
            continue

        child = Node(n, node, node.g + 1)
        heapq.heappush(open_list, child)

        # The flawed SMA* pruning mechanism
        if len(open_list) > MEM_LIMIT:
            open_list.sort(key=lambda x: x.f)
            open_list.pop()
```

### Experiment 6: Simplified Memory-Bounded A* (SMA*) for 8-puzzle problem
**Problem Statement**:<br>
Consider the 8-Puzzle problem, where tiles are arranged in a 3 \* 3 grid with one blank tile. The objective is to transform the initial configuration into the goal configuration using an optimal sequence of moves. Traditional A* search may run out of memory when the number of puzzle states increases significantly. Simplified Memory-Bounded A* (SMA*) is a variant of A* that uses a fixed memory limit by:
- Expanding best nodes first

**Task**:

**Constraints**:

- Heuristic must be admissible (Use Manhattan Distance)
- Environment is fully observable, deterministic, static, discrete, and single-agent
- SMA* must be implemented without built-in search libraries
- The algorithm must respect a predefined memory limit

```python
# ==========================================
# EXPERIMENT 6: "SMA*" 8-PUZZLE PROBLEM
# (Short/Truncated Version for Memorization)
# ==========================================
import heapq

MEM_LIMIT = 50 # Slightly higher limit since 8-puzzle has more branching
nodes_expanded = 0

# 1. State Definitions
init_state = (1, 2, 3,
              4, 0, 5,
              6, 7, 8)

goal_state = (1, 2, 3,
              4, 5, 6,
              7, 8, 0)

# 2. Heuristic: 8-Puzzle Manhattan Distance
def manhattan_8_puzzle(state, goal):
    dist = 0
    for tile in range(1, 9):
        curr_idx = state.index(tile)
        goal_idx = goal.index(tile)

        # Convert 1D index to 2D (row, col)
        curr_row, curr_col = curr_idx // 3, curr_idx % 3
        goal_row, goal_col = goal_idx // 3, goal_idx % 3

        dist += abs(curr_row - goal_row) + abs(curr_col - goal_col)
    return dist

# 3. The Node Class
class Node:
    def __init__(self, state, parent, g):
        self.state = state
        self.parent = parent
        self.g = g
        # f(n) = g(n) + h(n)
        self.f = g + manhattan_8_puzzle(state, goal_state)

    def __lt__(self, other):
        # heapq uses this to keep the lowest f-value at the top
        return self.f < other.f

# 4. Neighbor Generation (Tuple Swap Logic)
def get_neighbors(state):
    neighbors = []
    zero_idx = state.index(0)
    moves = []

    # Calculate valid index offsets
    if zero_idx > 2: moves.append(-3)              # UP
    if zero_idx < 6: moves.append(3)               # DOWN
    if zero_idx not in [0, 3, 6]: moves.append(-1) # LEFT
    if zero_idx not in [2, 5, 8]: moves.append(1)  # RIGHT

    for move in moves:
        new_idx = zero_idx + move
        new_state = list(state)
        # Swap 0 with the target tile
        new_state[zero_idx], new_state[new_idx] = new_state[new_idx], new_state[zero_idx]
        neighbors.append(tuple(new_state))

    return neighbors

# 5. The Main Loop
open_list = []
heapq.heappush(open_list, Node(init_state, None, 0))

while open_list:
    node = heapq.heappop(open_list)
    nodes_expanded += 1

    # Goal Test
    if node.state == goal_state:
        path = []
        while node:
            path.append(node.state)
            node = node.parent

        print(f"Path length: {len(path) - 1} steps")
        print("Path sequence:")
        for step in path[::-1]:
            print(f"{step[0:3]}\n{step[3:6]}\n{step[6:9]}\n")
        print(f"Nodes Expanded: {nodes_expanded}")
        break

    # Expand Node
    for n_state in get_neighbors(node.state):
        # Basic cycle avoidance (don't go immediately back to parent)
        if node.parent and n_state == node.parent.state:
            continue

        # Step cost is uniform (= 1)
        child = Node(n_state, node, node.g + 1)
        heapq.heappush(open_list, child)

        # The Simplified Pruning Mechanism
        if len(open_list) > MEM_LIMIT:
            # Sort the list by f-value in ascending order
            open_list.sort(key=lambda x: x.f)
            # Remove the last item (the node with the highest f-value)
            open_list.pop()
```

### Exercise 1: Hill-Climbing Search for Feature Selection in Predictive Modelling
**Problem Statement**:<br>
You are given a dataset (select a random dataset) with multiple input features and a target variable. Each state represents a subset of selected features. The goal is to find a feature subset that maximizes model accuracy.<br>

- **State Representation**<br>
    - A binary vector indicating whether a feature is selected (1) or not (0)
- **Initial State**<br>
    - A randomly generated feature subset
- **Neighbour Generation**<br>
    - Flip one bit (add/remove one feature)
- **Evaluation Function**<br>
    - Cross-validation accuracy of a simple classifier (e.g., Logistic Regression)
- **Algorithm Implementation**<br>
    - Implement steepest-ascent hill climbing
    - Stop when no neighbor improves the current solution
- **Experimentation**<br>
    - Run the algorithm multiple times with different random initial states

```python
# ==============================
# EXPERIMENT 1: HILL CLIMBING
# Feature Selection using Logistic Regression
# ==============================

import numpy as np
import random
from sklearn.datasets import load_breast_cancer
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import cross_val_score

# Load Dataset
data = load_breast_cancer()
X = data.data
y = data.target
n_features = X.shape[1]

# Evaluation Function
def evaluate(state):
    if sum(state) == 0:
        return 0
    selected_features = [i for i in range(n_features) if state[i] == 1]
    X_subset = X[:, selected_features]
    model = LogisticRegression(max_iter=5000)
    score = cross_val_score(model, X_subset, y, cv=5).mean()
    return score

# Generate Neighbors (Flip one bit)
def get_neighbors(state):
    neighbors = []
    for i in range(len(state)):
        neighbor = state.copy()
        neighbor[i] = 1 - neighbor[i]
        neighbors.append(neighbor)
    return neighbors

# Steepest-Ascent Hill Climbing
def hill_climbing():
    current = np.random.randint(0, 2, n_features)
    current_score = evaluate(current)

    while True:
        neighbors = get_neighbors(current)
        scores = [evaluate(neighbor) for neighbor in neighbors]
        best_neighbor = neighbors[np.argmax(scores)]
        best_score = max(scores)

        if best_score > current_score:
            current = best_neighbor
            current_score = best_score
        else:
            break

    return current, current_score

# Run multiple times
for run in range(3):
    solution, score = hill_climbing()
    print(f"Run {run+1}")
    print("Selected Features:", np.where(solution == 1)[0])
    print("Best Accuracy:", score)
    print("-"*50)
```

### Exercise 2: Simulated Annealing for Vehicle Routing Optimization
**Problem Statement**:<br>
A delivery vehicle must visit a set of locations and return to the depot. The order of visiting locations affects the total travel cost. Each state represents a permutation of delivery locations.

- **State Representation**<br>
    - A sequence representing the visiting order of delivery points
- **Initial State**<br>
    - A random permutation of locations
- **Neighbor Generation**<br>
    - Swap two locations in the route
- **Cost Function**<br>
    - Total travel distance of the route
- **Simulated Annealing Implementation**<br>
    - Implement SA using:
        - Temperature schedule T(t)
        - Acceptance probability: $P = e^{-\Delta/𝑇}$

```python
# ==============================
# EXPERIMENT 2: SIMULATED ANNEALING
# Vehicle Routing (TSP-like Problem)
# ==============================

import numpy as np
import random
import math

# Generate random locations (including depot at index 0)
num_locations = 10
locations = np.random.rand(num_locations, 2) * 100

# Distance Function
def total_distance(route):
    distance = 0
    for i in range(len(route) - 1):
        distance += np.linalg.norm(locations[route[i]] - locations[route[i+1]])
    distance += np.linalg.norm(locations[route[-1]] - locations[route[0]])
    return distance

# Neighbor by swapping two cities
def get_neighbor(route):
    new_route = route.copy()
    i, j = random.sample(range(len(route)), 2)
    new_route[i], new_route[j] = new_route[j], new_route[i]
    return new_route

# Simulated Annealing
def simulated_annealing():
    current = list(range(num_locations))
    random.shuffle(current)
    current_cost = total_distance(current)

    T = 1000
    cooling_rate = 0.995
    min_T = 1e-3

    while T > min_T:
        neighbor = get_neighbor(current)
        neighbor_cost = total_distance(neighbor)
        delta = neighbor_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = neighbor
            current_cost = neighbor_cost

        T *= cooling_rate

    return current, current_cost

best_route, best_cost = simulated_annealing()
print("Best Route:", best_route)
print("Minimum Distance:", best_cost)
```

### Exercise 3 (Revised): Genetic Algorithm for Autonomous Drone Path Planning
**Problem Statement**:<br>
A drone must travel from a start location to a destination in a 2D environment with obstacles. Each candidate solution represents a sequence of waypoints defining a flight path. The goal is to evolve paths that minimize travel distance while avoiding obstacles.

- **Chromosome Representation**<br>
    - A fixed-length sequence of (x, y) waypoints
- **Initial Population**<br>
    - Randomly generated feasible paths
- **Fitness Function**
    - Minimize:
        - Total path length
        - Obstacle collisions (high penalty)
- **Genetic Operators**
    - Selection: Tournament selection
    - Crossover: One-point crossover on waypoint sequence
    - Mutation: Random perturbation of waypoint coordinates
- **Constraints**
    - Waypoints must remain within map boundaries
    - Paths intersecting obstacles are penalized
- **Termination Condition**
    - Fixed number of generations or convergence

```python
# ==============================
# EXPERIMENT 3: GENETIC ALGORITHM
# Drone Path Planning
# ==============================

import numpy as np
import random

# Environment Settings
MAP_SIZE = 100
START = np.array([0, 0])
END = np.array([100, 100])
NUM_WAYPOINTS = 5
POP_SIZE = 30
GENERATIONS = 100
MUTATION_RATE = 0.1

# Obstacles (circles: center_x, center_y, radius)
obstacles = [
    (50, 50, 15),
    (70, 20, 10)
]

# Generate Random Path
def random_path():
    waypoints = np.random.rand(NUM_WAYPOINTS, 2) * MAP_SIZE
    return waypoints

# Fitness Function
def fitness(path):
    total_dist = 0
    points = np.vstack([START, path, END])

    for i in range(len(points)-1):
        total_dist += np.linalg.norm(points[i] - points[i+1])

    penalty = 0
    for (cx, cy, r) in obstacles:
        for point in path:
            if np.linalg.norm(point - np.array([cx, cy])) < r:
                penalty += 1000

    return total_dist + penalty

# Tournament Selection
def tournament_selection(pop):
    k = 3
    selected = random.sample(pop, k)
    selected.sort(key=lambda x: fitness(x))
    return selected[0]

# One-point Crossover
def crossover(parent1, parent2):
    point = random.randint(1, NUM_WAYPOINTS-1)
    child = np.vstack([parent1[:point], parent2[point:]])
    return child

# Mutation
def mutate(path):
    if random.random() < MUTATION_RATE:
        idx = random.randint(0, NUM_WAYPOINTS-1)
        path[idx] += np.random.randn(2) * 5
        path[idx] = np.clip(path[idx], 0, MAP_SIZE)
    return path

# Initialize Population
population = [random_path() for _ in range(POP_SIZE)]

# Evolution Process
for gen in range(GENERATIONS):
    new_population = []
    for _ in range(POP_SIZE):
        p1 = tournament_selection(population)
        p2 = tournament_selection(population)
        child = crossover(p1, p2)
        child = mutate(child)
        new_population.append(child)

    population = sorted(new_population, key=lambda x: fitness(x))[:POP_SIZE]

best_solution = min(population, key=lambda x: fitness(x))
print("Best Path Fitness:", fitness(best_solution))
print("Waypoints:\n", best_solution)
```
