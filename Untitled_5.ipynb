```python
from collections import defaultdict
with open("wordcount.txt", "r") as f:
  lines = f.readlines()
mapped = []
for line in lines:
  words = line.strip().split()
  for word in words:
    mapped.append((word.lower(), 1))
shuffled = defaultdict(list)
for key, value in mapped:
  shuffled[key].append(value)
reduced = {}
for key, values in shuffled.items():
  reduced[key] = sum(values)
print("Word count output:")
for word, count in reduced.items():
  print(word, count)
```

```python
import sys
for line in sys.stdin:
  line = line.strip()
  if not line:
    continue
  parts = line.split('\t', 1)
  if len(parts) != 2:
    continue
  doc_id, text = parts
  for word in text.split():
    word = word.strip().capitalize()
    print(f"{word}\t{doc_id}")
```

```python
import sys
current_word = None
docs = set()
for line in sys.stdin:
  line = line.strip()
  if not line:
    continue
  word, doc = line.split('\t', 1)
  if word != current_word:
    if current_word:
      print(f"{current_word}\t{','.join(sorted(docs))}")
    current_word = word
    docs = {doc}
  else:
    docs.add(doc)
if current_word:
  print(f"{current_word}\t{','.join(sorted(docs))}")
```

Lab 8, Q1 (Log Analysis)

```python
from pyspark import SparkContext
sc = SparkContext(appName = "LogAnalysis")
log_rdd = sc.textFile("hdfs://localhost:9000/user/server_log.txt")
for line in log_rdd.take(5):
  print(line)

status_codes = log_rdd.map(lambda line: line.split(" ")[8])
status_count = status_codes.map(lambda code: (code, 1)).reduceByKey(lambda a, b: a + b)
for item in status_count.collect():
  print(item)

ips = log_rdd.map(lambda line: line.split(" ")[0])
ip_count = ips.map(lambda ip: (ip, 1)).reduceByKey(lambda a, b: a + b).sortBy(lambda x: -x[1])
for item in ip_count.collect():
  print(item)

errors = status_codes.filter(lambda code: code.startswith('4') or code.startswith('5'))
for item in errors.collect():
  print(item)

error_count = errors.count()
print(error_count)

status_count.saveAsTextFile("hdfs://localhost:9000/output/status_count")
ip_count.saveAsTextFile("hdfs://localhost:9000/output/ip_count")
sc.stop()
```

```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.classification import LogisticRegression
from pyspark.ml.regression import LinearRegression
from pyspark.ml import Pipeline

spark = SparkSession.builder.appName("AttritionAnalysis").getOrCreate()
df = spark.read.csv("hdfs:///user/employee_attrition.csv", header = True, inferSchema = True)
df = df.dropna()
indexers = [
    StringIndexer(inputCol = "Department", outputCol = "Department_idx"),
    StringIndexer(inputCol = "Gender", outputCol = "Gender_idx"),
    StringIndexer(inputCol = "JobRole", outputCol = "JobRole_idx"),
    StringIndexer(inputCol = "Attrition", outputCol = "Attrition_idx")
]

pipeline = Pipeline(stages = indexers)
df = pipeline.fit(df).transform(df)

assembler = VectorAssembler(
    inputCols = ["Age", "MonthlyIncome", "Department_idx", "JobRole_idx", "Gender_idx"],
    outputCol = "features"
)

df = assembler.transform(df)
train, test = df.randomSplit([0.8, 0.2])
lr = LogisticRegression(featuresCol = "features", labelCol = "Attrition_idx")
model = lr.fit(train)
predictions = model.transform(test)
predictions.select("Attrition_idx", "prediction").show()
lr_reg = LinearRegression(featuresCol = "features", labelCol = "MonthlyIncome")
reg_model = lr_reg.fit(train)
reg_model.transform(test).select("MonthlyIncome", "prediction").show()
spark.stop()
```

Lab 9, Q2

```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import StringIndexer, VectorAssembler
from pyspark.ml.classification import RandomForestClassifier
from pyspark.ml.evaluation import MultiClassClassificationEvaluator
from pyspark.ml import Pipeline

spark = SparkSession.builder.appName("EmployeePromotionPrediction").getOrCreate()
df = spark.read.csv("hdfs://localhost:9000/user/employee_promotion.csv", header = True, inferSchema = True)
print("Initial Data: ")
df.show(5)
df.dropna()
indexers = [
    StringIndexer(inputCol = "department", outputCol = "department_idx"),
    StringIndexer(inputCol = "education", outputCol = "education_idx"),
    StringIndexer(inputCol = "gender", outputCol = "gender_idx")
]

pipeline = Pipeline(stages = indexers)
df = pipeline.fit(df).transform(df)
print("Promotion Distribution: ")
df.groupBy("is_promoted").count().show()
print("Statistical Summary:")
df.describe().show()

features = [
    "age",
    "length_of_service",
    "previous_year_rating",
    "KPIs_met",
    "awards_won",
    "avg_training_score",
    "department_idx",
    "education_idx",
    "gender_idx"
]

assembler = VectorAssembler(inputCols = features, outputCol = "features")
df = assembler.transform(df)

train, test = df.randomSplit([0.8, 0.2], seed = 42)
rf = RandomForestClassifier(
    featuresCol = "features",
    labelCol = "is_promoted",
    numTrees = 50
)

model = rf.fit(train)
predictions = model.transform(test)
print("Sample Predictions:")
predictions.select("is_promoted", "prediction", "probability").show(10)

evaluator = MulticlassClassificationEvaluator(
    labelCol = "is_promoted",
    predictionCol = "prediction",
    metricName = "accuracy"
)

accuracy = evaluator.evaluate(predictions)
print("Model Accuracy:", accuracy)

model.write().overwrite().save("hdfs:///user/230968126/lab9/employee_data/promotion_model")

print("Model saved successfully!")

from pyspark.ml.classification import RandomForestClassificationModel

loaded_model = RandomForestClassificationModel.load(
    "hdfs:///user/230968126/lab9/employee_data/promotion_model"
)

print("Loaded Model Predictions:")
loaded_model.transform(test).select("is_promoted", "prediction").show(5)

spark.stop()
```
